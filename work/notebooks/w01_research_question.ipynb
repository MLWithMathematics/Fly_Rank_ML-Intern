*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

1. What decision does this improve?

3. What does a wrong answer cost?

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Load: use the local repo copy if present, else pull the raw file directly
LOCAL_PATH = "data/raw/content_refresh_anonymized.csv"
RAW_URL = "https://raw.githubusercontent.com/MLWithMathematics/Fly_Rank_ML-Intern/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(LOCAL_PATH if os.path.exists(LOCAL_PATH) else RAW_URL)

# --- Number 1: scale + diversity of the population we'd be segmenting ---
print(f"Rows: {len(df):,} | Clients: {df['client_id'].nunique()} | Content types: {df['content_type'].nunique()}")

# --- Number 2: is there real behavioral spread, or is everything basically the same? ---
# High coefficient of variation on core engagement signals = content items behave very differently
spread_cols = ["ctr", "engagement_rate", "scroll_rate"]
cv = (df[spread_cols].std() / df[spread_cols].mean()).round(2)
print("Coefficient of variation (higher = more spread = more reason to segment):")
print(cv.to_string())

# --- Number 3: does a quick clustering pass find real structure, or just noise? ---
# A handful of numeric behavioral features, has_* flags instead of blind fillna
num_cols = ["ctr", "avg_position", "engagement_rate", "scroll_rate",
            "pageviews_90d", "sessions_90d", "content_age_days"]
X = df[num_cols].copy()
for c in X.columns[X.isna().any()]:
    X[f"has_{c}"] = X[c].notna().astype(int)
    X[c] = X[c].fillna(X[c].median())

X_scaled = StandardScaler().fit_transform(X)
sample = X_scaled[np.random.RandomState(42).choice(len(X_scaled), 5000, replace=False)]
labels = KMeans(n_clusters=4, n_init=10, random_state=42).fit_predict(sample)
print(f"Silhouette score (k=4, sample of 5,000 rows): {silhouette_score(sample, labels):.3f}")
```
